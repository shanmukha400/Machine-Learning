```python
# PRACTICE-12: Write a program to implement Categorical Encoding.

import pandas as pd

from sklearn.preprocessing import LabelEncoder

# Load Dataset

data = pd.read_csv("Student_ML_Lab_Dataset.csv")

# Select Required Column

df = data[["Department"]].copy()

print("Original Dataset")

print(df)

# Label Encoding

encoder = LabelEncoder()

df["Department_Encoded"] = encoder.fit_transform(df["Department"])

print("\nEncoded Dataset")

print(df)

print("\nCategory Mapping")

for category, value in zip(encoder.classes_, encoder.transform(encoder.classes_)):

    print(category, "->", value)
```
