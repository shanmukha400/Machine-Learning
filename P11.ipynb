```python
# PRACTICE-11: Write a program to implement One-hot Encoding.

import pandas as pd

# Load Dataset

data = pd.read_csv("Student_ML_Lab_Dataset.csv")

# Select Required Columns

df = data[["Gender", "Department", "City"]]

print("Original Dataset")

print(df)

# One-Hot Encoding

encoded_df = pd.get_dummies(df, columns=["Gender", "Department", "City"])

print("\nOne-Hot Encoded Dataset")

print(encoded_df)
```
